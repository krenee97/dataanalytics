## Lab 3. Tax Functions 
this notebook calculates taxes based on gross pay and withholding code.

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        rate = 0.18 - ((withholding_code - 4) * 0.005)
    return gross_pay * rate
```

```python
# Person 1: gross $750, withholding code 0
print(f"Person 1 SS Tax: ${round(get_soc_sec_tax(750), 2)}")
print(f"Person 1 Medicare Tax: ${round(get_medicare_tax(750), 2)}")
print(f"Person 1 Federal Tax: ${round(get_federal_tax(750), 0), 2}")

# Person 2: gross $1500, withholding code 2
print(f"person 2 SS Tax: ${round(get_soc_sec_tax(1550), 2)}")
print(f"Person 2 Medicare Tax: ${round(get_medicare_tax(1550), 2)}")
print(f"Person 2 Federal Tax ${round(get_federal_tax(1550, 2), 2)}")

# Person 3: gross $110, withholding code 5
print(f"Person 3 SS Tax: ${round(get_soc_sec_tax(1100), 2)}")
print(f"Person 3 Medicare Tax: ${round(get_medicare_tax(1100), 2)}")
print(f"Person 3 Federal Tax: ${round(get_federal_tax(1100, 5), 2)}")
```
