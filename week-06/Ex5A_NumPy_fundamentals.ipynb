## Array Basics

```python
import numpy as np
```

```python
fruits = ['apple', 'banana', 'cherry', 'date']
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)
# dtype U6 means Unicode string, the number is the max string length
```

```python
arr-2 = np.array([(10, 11, 12), (13, 14, 15)])
print(arr_2)
```

```python
print(arr_2.T)
# Transposing does not alter the original arr_2
print(arr_2)
```

```python
arr_3 = np.array([(1, 3, 5), (2, 4, 6)])
print(arr_3)
```

```python
arr_4 =np.array([arr_3[0, 2], arr_3[1, 1]])
print(arr_4)
```
