## Lab 2. Working with datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
print(f"Today is {today.strftime('%A, %B %d, %Y')}")
```

# My Birthday

```python
birthday = datetime.date(1990, 6, 15)
print(f"My birthday is {birthday.strftime('%m/%d/%Y')}")
```

# Next 90 days

```python
ninety_d = today = datetime.timedelta(days=90)
print(f"90 days from today is {ninety_d.strftime('%B %d, %Y')}")
```

# Time for Dinner

```python
dinner_time = datetime.datetime.now()
print(f"Lets's meet for dinner at {dinner_time.strftime('%I:%M %p')}")
```
